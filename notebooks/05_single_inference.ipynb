## 🔄 Single Sample Prediction (Real-Time Style)

# Imports

```python
import pandas as pd
import joblib
from tensorflow.keras.models import load_model

import sys
sys.path.append("../")  # Adjust the path if src is elsewhere

from src.utils.data_utils import preprocess_input
```

# Code

```python
# Load model and scaler
model = load_model("../models/cnn_rnn_model.h5")
scaler = joblib.load("../models/scaler.pkl")

# Load one row from mock sensor data
df_new = pd.read_csv("../data/robot_sensor_data.csv")
sample = df_new.iloc[0:1]  # take the first row as DataFrame

# Preprocess sample with existing scaler
X_new, _ = preprocess_input(sample, scaler)

# Run prediction
pred = model.predict(X_new)[0][0]
label = "FAILURE" if pred > 0.5 else "OK"

print(f"🔍 Prediction for first row: {label} (prob: {pred:.2f})")
```
